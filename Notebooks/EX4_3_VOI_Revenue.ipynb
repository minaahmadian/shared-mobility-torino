```python
import pandas as pd

# path to your file
file_path = "../Data/df_all_voi.csv"

# read data
df = pd.read_csv(file_path)

# show basic info
print("Shape of dataset:", df.shape)

# show columns
print("\nColumns:")
print(df.columns)

# preview data
df.head()
```

```python
df_dup = df[df.duplicated(subset=['Identificativo noleggio'], keep=False)]

print("Number of duplicated trip IDs:", df_dup.shape)
df_dup.head()
```

```python
# 1. remove duplicates
df = df.drop_duplicates()

print("After removing duplicates:", df.shape)


# 2. remove invalid durations (<= 0)
df = df[df['Tempo Tot'] > 0]

print("After removing zero/negative durations:", df.shape)


# 3. check duration stats
print("\nDuration stats (Tempo Tot):")
print(df['Tempo Tot'].describe())
```

```python
# ===== 1. Check duplicates (full row) =====
print("Full duplicate rows:", df.duplicated().sum())


# ===== 2. Check duplicate Trip IDs =====
dup_ids = df['Identificativo noleggio'].duplicated().sum()
print("Duplicate trip IDs:", dup_ids)


# ===== 3. Check missing values =====
print("\nMissing values per column:")
print(df.isnull().sum())


# ===== 4. Check time consistency =====
df['Data inizio corsa'] = pd.to_datetime(df['Data inizio corsa'])
df['Data fine corsa'] = pd.to_datetime(df['Data fine corsa'])

wrong_time = df[df['Data fine corsa'] < df['Data inizio corsa']]
print("\nTrips with wrong time (end < start):", wrong_time.shape[0])


# ===== 5. Check duration consistency =====
# duration from timestamps (in minutes)
df['duration_calc'] = (df['Data fine corsa'] - df['Data inizio corsa']).dt.total_seconds() / 60

# difference between given and calculated
df['duration_diff'] = abs(df['Tempo Tot'] - df['duration_calc'])

print("\nDuration difference stats:")
print(df['duration_diff'].describe())


# ===== 6. Check extreme values =====
print("\nVery long trips (>60 min):", df[df['Tempo Tot'] > 60].shape[0])
print("Very short trips (<1 min):", df[df['Tempo Tot'] < 1].shape[0])
```

```python
# remove duplicate trip IDs (keep first occurrence)
df = df.drop_duplicates(subset=['Identificativo noleggio'])

print("After removing duplicate trip IDs:", df.shape)
```

```python
print("Remaining duplicate IDs:", df['Identificativo noleggio'].duplicated().sum())
```

```python
# correct tariff from VOI app
unlock_fee = 1.0
price_per_min = 0.24

# recompute revenue
df['revenue_unlock'] = unlock_fee
df['revenue_time'] = df['Tempo Tot'] * price_per_min
df['revenue_total'] = df['revenue_unlock'] + df['revenue_time']

# totals
total_revenue = df['revenue_total'].sum()
unlock_total = df['revenue_unlock'].sum()
time_total = df['revenue_time'].sum()

print("Total revenue:", round(total_revenue, 2))
print("Unlock revenue:", round(unlock_total, 2))
print("Time revenue:", round(time_total, 2))

print("\nShares:")
print("Unlock %:", round(unlock_total / total_revenue * 100, 2))
print("Time %:", round(time_total / total_revenue * 100, 2))
```

```python
# average revenue per trip
print("Average revenue per trip:", round(df['revenue_total'].mean(), 2))


# distribution
df['revenue_total'].describe()
```

```python
n_scooters = df['Targa veicolo'].nunique()
print("Number of scooters:", n_scooters)
```

```python
fixed_cost_per_scooter = 400

total_fixed_cost = n_scooters * fixed_cost_per_scooter

print("Total fixed cost:", total_fixed_cost)
```

```python
cost_per_trip = 0.15  # electricity + maintenance

total_variable_cost = len(df) * cost_per_trip

print("Total variable cost:", total_variable_cost)
```

```python
total_cost = total_fixed_cost + total_variable_cost

print("Total cost:", total_cost)
```

```python
profit = total_revenue - total_cost

print("Profit:", profit)
```

```python
# tax rate (Italy approx)
tax_rate = 0.28

# profit before tax
profit_before_tax = total_revenue - total_cost

# tax amount
tax = profit_before_tax * tax_rate

# profit after tax
profit_after_tax = profit_before_tax - tax

print("Profit before tax:", round(profit_before_tax, 2))
print("Tax:", round(tax, 2))
print("Profit after tax:", round(profit_after_tax, 2))
```

```python

import plotly.express as px
import plotly.graph_objects as go
```

```python
df_filtered = df[df['revenue_total'] <= 15]

fig = px.histogram(
    df_filtered,
    x="revenue_total",
    nbins=50,
    title="Distribution of Revenue per Trip (VOI) - Filtered (≤15€)",
)

fig.update_layout(
    xaxis_title="Revenue per Trip (€)",
    yaxis_title="Number of Trips"
)

fig.show()
```

```python
fig = px.scatter(
    df.sample(5000),  # برای سرعت
    x="Tempo Tot",
    y="revenue_total",
    title="Revenue vs Trip Duration",
    opacity=0.5
)

fig.update_layout(
    xaxis_title="Duration (minutes)",
    yaxis_title="Revenue (€)"
)

fig.show()
```

```python
labels = ["Unlock Revenue", "Time-based Revenue"]
values = [unlock_total, time_total]

fig = go.Figure(data=[go.Pie(labels=labels, values=values, hole=0.4)])

fig.update_layout(title="Revenue Composition (VOI)")

fig.show()
```

```python
revenue_by_hour = df.groupby('hour')['revenue_total'].sum().reset_index()

fig = px.line(
    revenue_by_hour,
    x='hour',
    y='revenue_total',
    title="Revenue by Hour of Day"
)

fig.update_layout(
    xaxis_title="Hour",
    yaxis_title="Total Revenue (€)"
)

fig.show()
```

```python
rev_per_scooter = df.groupby('Targa veicolo')['revenue_total'].sum().reset_index()

fig = px.histogram(
    rev_per_scooter,
    x='revenue_total',
    nbins=40,
    title="Revenue per Scooter Distribution"
)

fig.update_layout(
    xaxis_title="Revenue per Scooter (€)",
    yaxis_title="Number of Scooters"
)

fig.show()
```
