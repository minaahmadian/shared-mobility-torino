```python
import pandas as pd

# path to your file
file_path = "../Data/df_all_bird.csv"

# read data
df = pd.read_csv(file_path)

# show basic info
print("Shape of dataset:", df.shape)

# show columns
print("\nColumns:")
print(df.columns)

# preview data
df.head()
```

```python
print(df.shape)
df.info()
df.describe()
```

```python
# حذف duration های صفر یا منفی
df = df[df['DURATA_MIN'] > 0]

# حذف outlier های خیلی غیرواقعی (مثلاً > 120 دقیقه)
df = df[df['DURATA_MIN'] < 120]

# حذف سرعت‌های غیرمنطقی (مثلاً > 50 km/h)
df = df[df['SPEED_KMH'] < 50]
```

```python
df.isna().sum()
```

```python
df.duplicated().sum()
df.duplicated(subset=['ID_VEICOLO','DATAORA_INIZIO','DATAORA_FINE']).sum()
```

```python
# Remove useless columns
df = df.drop(columns=['PERCORSO', 'RISERVATO'])

# Remove unrealistic durations
df = df[(df['DURATA_MIN'] > 1) & (df['DURATA_MIN'] < 120)]

# Remove GPS outliers
df = df[df['HAV_DIST_KM'] < 20]

# Optional: remove very low speeds (noise)
df = df[df['SPEED_KMH'] > 3]
```

```python
df.shape
df.describe()
```

```python
def bird_revenue(duration):
    
    # Bird+ users (60%)
    if duration <= 10:
        rev_plus = 1.5
    else:
        rev_plus = 1.5 + (duration - 10) * 0.20   # slightly discounted rate
    
    # Normal users (40%)
    rev_normal = 1 + duration * 0.25
    
    # Weighted average
    return 0.6 * rev_plus + 0.4 * rev_normal
```

```python
df['rev_bird'] = df['DURATA_MIN'].apply(bird_revenue)

total_revenue = df['rev_bird'].sum()

print("Total Revenue (Bird):", round(total_revenue, 2))
```

```python
num_users = len(df) / 20   # same assumption as Lime
subscription_revenue = num_users * 1   # €1/week

print("Subscription revenue:", round(subscription_revenue, 2))
```

```python
num_scooters = df['ID_VEICOLO'].nunique()
print("Number of scooters:", num_scooters)
```

```python
fixed_cost_per_scooter = 400  # € per scooter (purchase + depreciation + insurance)

total_fixed_cost = num_scooters * fixed_cost_per_scooter

print("Total fixed cost:", total_fixed_cost)
```

```python
variable_cost_per_min = 0.06  # € per minute (electricity + maintenance)

total_variable_cost = (df['DURATA_MIN'] * variable_cost_per_min).sum()

print("Total variable cost:", round(total_variable_cost, 2))
```

```python
total_cost = total_fixed_cost + total_variable_cost

print("Total cost:", round(total_cost, 2))
```

```python
total_final_revenue = total_revenue + subscription_revenue

profit_before_tax = total_final_revenue - total_cost

print("Profit before tax:", round(profit_before_tax, 2))
```

```python
tax_rate = 0.28

tax = max(profit_before_tax, 0) * tax_rate

print("Tax:", round(tax, 2))
```

```python
profit_after_tax = profit_before_tax - tax

print("Profit after tax:", round(profit_after_tax, 2))
```

```python
import plotly.express as px

rev_hour = df.groupby('hour')['rev_bird'].sum().reset_index()

fig = px.line(
    rev_hour,
    x='hour',
    y='rev_bird',
    title="Revenue by Hour of Day (Bird)"
)

fig.update_layout(
    xaxis_title="Hour",
    yaxis_title="Total Revenue (€)"
)

fig.show()
```

```python
avg_rev_hour = df.groupby('hour')['rev_bird'].mean().reset_index()

fig = px.line(
    avg_rev_hour,
    x='hour',
    y='rev_bird',
    title="Average Revenue per Trip by Hour (Bird)"
)

fig.update_layout(
    xaxis_title="Hour",
    yaxis_title="Average Revenue (€)"
)

fig.show()
```

```python
import plotly.express as px

rev_per_scooter = df.groupby('ID_VEICOLO')['rev_bird'].sum().reset_index()

fig = px.histogram(
    rev_per_scooter,
    x='rev_bird',
    nbins=50,
    title="Revenue per Scooter Distribution (Bird)"
)

fig.update_layout(
    xaxis_title="Revenue per Scooter (€)",
    yaxis_title="Number of Scooters"
)

fig.show()
```
