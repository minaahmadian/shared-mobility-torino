```python
import pandas as pd

data = {
    'Company': ['VOI', 'Lime', 'Bird'],

    # Revenue
    'Total_Revenue': [
        729176.5,        # VOI
        2624617.57,      # Lime
        2811181.77         # Bird (بر اساس scale و نمودارها)
    ],

    # Cost
    'Total_Cost': [
        611965.4,        # VOI
        1849117.94,      # Lime
        1811865.18          # Bird
    ],

    # Profit before tax
    'Profit_Before_Tax': [
        117211.1,        # VOI
        775499.63,       # Lime
        999316.59           # Bird
    ],

    # Tax (28%)
    'Tax': [
        32819.11,        # VOI
        217139.90,       # Lime
        279808.65           # Bird
    ],

    # Profit after tax
    'Profit_After_Tax': [
        84391.99,        # VOI
        558359.73,       # Lime
        719507.95           # Bird
    ],

    # Fleet size
    'Num_Scooters': [
        1445,            # VOI
        2399,            # Lime
        2816             # Bird (approx from dataset scale)
    ]
}

df_compare = pd.DataFrame(data)
df_compare
```

```python
df_compare[['Company',
            'Revenue_per_Scooter',
            'Cost_per_Scooter',
            'Profit_per_Scooter']] = pd.DataFrame({
    'Company': df_compare['Company'],
    'Revenue_per_Scooter': df_compare['Total_Revenue'] / df_compare['Num_Scooters'],
    'Cost_per_Scooter': df_compare['Total_Cost'] / df_compare['Num_Scooters'],
    'Profit_per_Scooter': df_compare['Profit_After_Tax'] / df_compare['Num_Scooters']
})[['Company',
    'Revenue_per_Scooter',
    'Cost_per_Scooter',
    'Profit_per_Scooter']]

df_compare[['Company',
            'Revenue_per_Scooter',
            'Cost_per_Scooter',
            'Profit_per_Scooter']].round(2)
```

```python
import plotly.express as px

df_melt = df_compare.melt(
    id_vars='Company',
    value_vars=['Revenue_per_Scooter', 'Cost_per_Scooter', 'Profit_per_Scooter'],
    var_name='Metric',
    value_name='Value'
)

fig = px.bar(
    df_melt,
    x='Company',
    y='Value',
    color='Metric',
    barmode='group',
    title='Per-Scooter Performance Comparison'
)

fig.show()
```

```python
df_compare['Profit_Margin'] = df_compare['Profit_After_Tax'] / df_compare['Total_Revenue']

import plotly.express as px

fig = px.bar(
    df_compare,
    x='Company',
    y='Profit_Margin',
    title='Profit Margin Comparison',
    text_auto='.2%'
)

fig.show()
```

```python
import plotly.graph_objects as go

fig = go.Figure()

# Revenue
fig.add_trace(go.Bar(
    x=df_compare['Company'],
    y=df_compare['Total_Revenue'],
    name='Revenue'
))

# Profit
fig.add_trace(go.Bar(
    x=df_compare['Company'],
    y=df_compare['Profit_After_Tax'],
    name='Profit'
))

fig.update_layout(
    title='Revenue vs Profit Comparison',
    barmode='group'
)

fig.show()
```

```python
df_compare['Cost_Ratio'] = df_compare['Total_Cost'] / df_compare['Total_Revenue']

import plotly.express as px

fig = px.bar(
    df_compare,
    x='Company',
    y='Cost_Ratio',
    title='Cost-to-Revenue Ratio Comparison',
    text_auto='.2%'
)

fig.show()
```

```python
import plotly.express as px

fig = px.scatter(
    df_compare,
    x='Num_Scooters',
    y='Profit_After_Tax',
    text='Company',
    size='Total_Revenue',
    title='Fleet Size vs Profit (Scaling Effect)'
)

fig.update_traces(textposition='top center')

fig.show()
```

```python
import plotly.express as px

fig = px.scatter(
    df_compare,
    x='Total_Cost',
    y='Total_Revenue',
    text='Company',
    size='Profit_After_Tax',
    title='Revenue vs Cost (Efficiency & Break-even)'
)

# خط break-even (Revenue = Cost)
fig.add_shape(
    type='line',
    x0=df_compare['Total_Cost'].min(),
    y0=df_compare['Total_Cost'].min(),
    x1=df_compare['Total_Cost'].max(),
    y1=df_compare['Total_Cost'].max(),
    line=dict(dash='dash')
)

fig.update_traces(textposition='top center')

fig.show()
```
