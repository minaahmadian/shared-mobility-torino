```python
import pandas as pd
import geopandas as gpd
from shapely.geometry import Point
lime = pd.read_csv("../data/df_all_lime.csv")
voi= pd.read_csv("../data/df_all_voi.csv")
bird = pd.read_csv("../data/df_all_bird.csv")
```

```python
print("Lime columns:\n", lime.columns)
print("\nVoi columns:\n", voi.columns)
print("\nBird columns:\n", bird.columns)
```

```python
lime = lime.rename(columns={
    "LATITUDINE_INIZIO_CORSA": "start_lat",
    "LONGITUTIDE_INIZIO_CORSA": "start_lon",
    "LATITUDINE_FINE_CORSA": "end_lat",
    "LONGITUTIDE_FINE_CORSA": "end_lon",
    "DATAORA_INIZIO": "start_time",
    "DATAORA_FINE": "end_time"
})
```

```python
bird = bird.rename(columns={
    "LATITUDINE_INIZIO_CORSA": "start_lat",
    "LONGITUTIDE_INIZIO_CORSA": "start_lon",
    "LATITUDINE_FINE_CORSA": "end_lat",
    "LONGITUTIDE_FINE_CORSA": "end_lon",
    "DATAORA_INIZIO": "start_time",
    "DATAORA_FINE": "end_time"
})
```

```python
voi = voi.rename(columns={
    "Lat inizio corsa_coordinate": "start_lat",
    "Lon inizio corsa_coordinate": "start_lon",
    "Lat fine corsa_coordinate": "end_lat",
    "Lon fine corsa_coordinate": "end_lon",
    "Data inizio corsa": "start_time",
    "Data fine corsa": "end_time"
})
```

```python
print("Lime:", lime[["start_lat","start_lon","end_lat","end_lon","start_time","end_time"]].head())
print("\nVoi:", voi[["start_lat","start_lon","end_lat","end_lon","start_time","end_time"]].head())
print("\nBird:", bird[["start_lat","start_lon","end_lat","end_lon","start_time","end_time"]].head())
```

```python
lime["start_time"] = pd.to_datetime(lime["start_time"])
lime["end_time"]   = pd.to_datetime(lime["end_time"])

voi["start_time"] = pd.to_datetime(voi["start_time"])
voi["end_time"]   = pd.to_datetime(voi["end_time"])

bird["start_time"] = pd.to_datetime(bird["start_time"])
bird["end_time"]   = pd.to_datetime(bird["end_time"])
```

```python
print(lime["start_time"].dtype)
print(voi["start_time"].dtype)
print(bird["start_time"].dtype)
```

```python
lime["operator"] = "Lime"
voi["operator"]  = "Voi"
bird["operator"] = "Bird"
```

```python
print(lime["operator"].unique())
print(voi["operator"].unique())
print(bird["operator"].unique())
```

```python
df_all = pd.concat([lime, voi, bird], ignore_index=True)
```

```python
print(df_all.shape)
print(df_all["operator"].value_counts())
```

```python
df_all["year_month"] = df_all["start_time"].dt.to_period("M")

print(df_all["year_month"].value_counts().sort_index())
```

```python
print("Lime:\n", 
      lime["start_time"].dt.to_period("M").value_counts().sort_index())


```

```python
print("\nBird:\n", 
      bird["start_time"].dt.to_period("M").value_counts().sort_index())
```

```python
print("\nVoi:\n", 
      voi["start_time"].dt.to_period("M").value_counts().sort_index())
```

```python
df_oct = df_all[df_all["start_time"].dt.to_period("M") == "2024-10"].copy()
```

```python
print(df_oct.shape)
print(df_oct["operator"].value_counts())
```

```python
import geopandas as gpd
from shapely.geometry import Point

gdf_oct = gpd.GeoDataFrame(
    df_oct,
    geometry=gpd.points_from_xy(df_oct["start_lon"], df_oct["start_lat"]),
    crs="EPSG:4326"
)
```

```python
print(gdf_oct.head())
print(gdf_oct.crs)
```

```python
gdf_oct = gdf_oct.loc[:, ~gdf_oct.columns.duplicated()]
```

```python
gdf_oct = gdf_oct[["start_lat", "start_lon", "geometry"]]
```

```python
gdf_end = gpd.GeoDataFrame(
    df_oct,
    geometry=gpd.points_from_xy(df_oct["end_lon"], df_oct["end_lat"]),
    crs="EPSG:4326"
)
```
