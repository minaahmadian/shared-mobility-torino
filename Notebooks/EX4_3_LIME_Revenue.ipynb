```python
import pandas as pd

# path to your file
file_path = "../Data/df_all_lime.csv"

# read data
df = pd.read_csv(file_path)

# show basic info
print("Shape of dataset:", df.shape)

# show columns
print("\nColumns:")
print(df.columns)

# preview data
df.head()
```

```python
# check missing values
missing = df.isnull().sum().sort_values(ascending=False)
missing_percent = (missing / len(df)) * 100

missing_df = pd.DataFrame({
    'Missing_Count': missing,
    'Missing_Percent': missing_percent
})

missing_df
```

```python
# check duplicates (full row)
duplicates = df.duplicated().sum()
print("Number of duplicate rows:", duplicates)
```

```python
# check duplicates based on key trip features
dup_trip = df.duplicated(subset=[
    'ID_VEICOLO', 'DATAORA_INIZIO', 'DATAORA_FINE'
]).sum()

print("Duplicate trips (based on vehicle + time):", dup_trip)
```

```python
df.dtypes
```

```python
df['DATAORA_INIZIO'] = pd.to_datetime(df['DATAORA_INIZIO'])
df['DATAORA_FINE'] = pd.to_datetime(df['DATAORA_FINE'])
```

```python
# remove invalid durations
df = df[df['DURATA_MIN'] > 0]

# remove zero distance
df = df[df['DISTANZA_KM'] > 0]

# remove unrealistic speed
df = df[df['SPEED_KMH'] <= 50]

# remove time inconsistencies
df = df[df['DATAORA_FINE'] >= df['DATAORA_INIZIO']]
```

```python
print("Final shape:", df.shape)
```

```python
print("Min duration:", df['DURATA_MIN'].min())
print("Min distance:", df['DISTANZA_KM'].min())
print("Max speed:", df['SPEED_KMH'].max())

print("Invalid time rows:", (df['DATAORA_FINE'] < df['DATAORA_INIZIO']).sum())
```

```python
def lime_revenue(duration):
    
    if duration <= 5:
        return 1.0
    
    elif duration <= 20:
        return 1.8
    
    else:
        return 1.8 + (duration - 20) * 0.14
```

```python
df['rev_lime_prime'] = df['DURATA_MIN'].apply(lime_revenue)

total_rev = df['rev_lime_prime'].sum()

print("Total Revenue (LimePrime - rides only):", round(total_rev, 2))
```

```python
num_users = len(df) / 20   # فرض 20 trip per user

subscription_revenue = num_users * 2.99

print("Estimated Subscription Revenue:", round(subscription_revenue, 2))
```

```python
total_final = total_rev + subscription_revenue

print("Total Revenue (LimePrime total):", round(total_final, 2))
```

```python
num_scooters = df['ID_VEICOLO'].nunique()

print("Number of scooters:", num_scooters)
```

```python
fixed_cost_per_scooter = 400

total_fixed_cost = num_scooters * fixed_cost_per_scooter

print("Total fixed cost:", total_fixed_cost)
```

```python
variable_cost_per_min = 0.06

total_variable_cost = (df['DURATA_MIN'] * variable_cost_per_min).sum()

print("Total variable cost:", round(total_variable_cost, 2))
```

```python
total_cost = total_fixed_cost + total_variable_cost

print("Total cost:", round(total_cost, 2))
```

```python
profit_before_tax = total_final - total_cost

print("Profit before tax:", round(profit_before_tax, 2))
```

```python
tax_rate = 0.28

tax = max(profit_before_tax, 0) * tax_rate

print("Tax:", round(tax, 2))
```

```python
profit_after_tax = profit_before_tax - tax

print("Profit after tax:", round(profit_after_tax, 2))
```

```python
import plotly.express as px

# حذف outlierها برای خوانایی بهتر
df_filtered = df[df['rev_lime_prime'] <= 4]

fig = px.histogram(
    df_filtered,
    x="rev_lime_prime",
    nbins=40,
    title="Distribution of Revenue per Trip (LimePrime - Filtered ≤4€)"
)

fig.update_layout(
    xaxis_title="Revenue per Trip (€)",
    yaxis_title="Number of Trips"
)

fig.show()
```

```python
fig = px.scatter(
    df.sample(5000),
    x="DURATA_MIN",
    y="rev_lime_prime",
    title="Revenue vs Trip Duration (Lime)",
    opacity=0.5
)

fig.update_layout(
    xaxis_title="Duration (minutes)",
    yaxis_title="Revenue (€)"
)

fig.show()
```

```python
revenue_by_hour = df.groupby('HOUR')['rev_lime_prime'].sum().reset_index()

fig = px.line(
    revenue_by_hour,
    x='HOUR',
    y='rev_lime_prime',
    title="Revenue by Hour of Day (Lime)"
)

fig.update_layout(
    xaxis_title="Hour",
    yaxis_title="Total Revenue (€)"
)

fig.show()
```

```python
rev_per_scooter = df.groupby('ID_VEICOLO')['rev_lime_prime'].sum().reset_index()

fig = px.histogram(
    rev_per_scooter,
    x='rev_lime_prime',
    nbins=40,
    title="Revenue per Scooter Distribution (Lime)"
)

fig.update_layout(
    xaxis_title="Revenue per Scooter (€)",
    yaxis_title="Number of Scooters"
)

fig.show()
```

```python
avg_rev_hour = df.groupby('HOUR')['rev_lime_prime'].mean().reset_index()

fig = px.line(
    avg_rev_hour,
    x='HOUR',
    y='rev_lime_prime',
    title="Average Revenue per Trip by Hour (Lime)"
)

fig.update_layout(
    xaxis_title="Hour",
    yaxis_title="Avg Revenue (€)"
)

fig.show()
```
